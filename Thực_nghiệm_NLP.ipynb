```python
import fitz  # PyMuPDF

documents = []

for filename in uploaded.keys():

    # Mở từng tài liệu PDF
    doc = fitz.open(filename)

    print("Đang xử lý:", filename)

    # Duyệt từng trang trong tài liệu
    for page_number, page in enumerate(doc, start=1):

        text = page.get_text("text")

        documents.append({
            "document_id": filename,
            "page": page_number,
            "text": text
        })

    doc.close()

print("Tổng số trang đã trích xuất:", len(documents))
```

```python
import re

def clean_text(text):

    # Loại bỏ ký tự không cần thiết
    text = text.replace("\u00ad", "")

    # Chuẩn hóa khoảng trắng và tab
    text = re.sub(r"[ \t]+", " ", text)

    # Chuẩn hóa xuống dòng
    text = re.sub(r"\n+", "\n", text)

    # Xóa khoảng trắng đầu và cuối
    text = text.strip()

    return text
```

Chia văn bản thành chunks

```python
def split_into_chunks(text, max_words=250, overlap=40):

    # Tách thành các đoạn văn
    paragraphs = [
        p.strip()
        for p in text.split("\n")
        if p.strip()
    ]

    chunks = []
    current_words = []

    for paragraph in paragraphs:

        words = paragraph.split()

        # Nếu một đoạn dài hơn giới hạn
        # thì chia nhỏ đoạn đó
        if len(words) > max_words:

            if current_words:
                chunks.append(" ".join(current_words))
                current_words = []

            start = 0

            while start < len(words):

                part = words[start:start + max_words]
                chunks.append(" ".join(part))

                if start + max_words >= len(words):
                    break

                start += max_words - overlap

        # Nếu thêm đoạn mới vượt giới hạn
        elif len(current_words) + len(words) > max_words:

            chunks.append(" ".join(current_words))

            # Giữ lại phần cuối để tạo overlap
            current_words = (
                current_words[-overlap:]
                if overlap > 0 else []
            )

            current_words.extend(words)

        else:
            current_words.extend(words)

    # Lưu phần còn lại
    if current_words:
        chunks.append(" ".join(current_words))

    return chunks
```

```python
all_chunks = []

for item in documents:

    chunks = split_into_chunks(item["text"])

    for i, chunk in enumerate(chunks, start=1):

        all_chunks.append({
            "document_id": item["document_id"],
            "page": item["page"],
            "chunk_id": f"{item['document_id']}_P{item['page']}_C{i}",
            "text": chunk
        })

print("Tổng số chunks:", len(all_chunks))
```

```python
# Kiểm tra cấu trúc
required_columns = ["document_id", "page", "chunk_id", "text"]

for col in required_columns:
    if col not in df.columns:
        raise ValueError(f"Thiếu cột: {col}")

df = df.dropna(subset=["text"]).copy()

# Chuyển dữ liệu về kiểu chuỗi
df["text"] = df["text"].astype(str)

# Tạo ID duy nhất cho mỗi chunk
df["chunk_id"] = df["chunk_id"].astype(str)

print("Số lượng chunks:", len(df))
print(df.head())
```

```python
def normalize_text(text):
    text = str(text)

    # Chuẩn hóa Unicode
    text = unicodedata.normalize("NFC", text)

    # Xử lý các ligature Unicode chuẩn
    text = text.replace("\ufb01", "fi")
    text = text.replace("\ufb02", "fl")

    # Bỏ ký tự soft hyphen
    text = text.replace("\u00ad", "")

    # Chuẩn hóa khoảng trắng
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()


df["text_normalized"] = df["text"].apply(normalize_text)

print(df[["text", "text_normalized"]].head(5))
```

```python
from underthesea import word_tokenize, text_normalize

def inspect_chunk(text):
    text = str(text)

    issues = []
    suggestions = []

    # 1. Kiểm tra chuỗi chữ cái quá dài không có khoảng trắng
    long_strings = re.findall(
        r"[A-Za-zÀ-ỹĐđ]{18,}",
        text
    )

    for item in long_strings:
        issues.append("Chuỗi chữ dài bất thường")
        suggestions.append(item)

    # 2. Kiểm tra ký tự Unicode thuộc vùng riêng hoặc ký tự thay thế
    unusual_chars = []

    for char in text:
        code = ord(char)

        if (
            0xE000 <= code <= 0xF8FF
            or char == "\uFFFD"
        ):
            unusual_chars.append(
                f"{char} (U+{code:04X})"
            )

    if unusual_chars:
        issues.append("Ký tự Unicode bất thường")
        suggestions.extend(unusual_chars)

    # 3. Kiểm tra dấu hiệu mã hóa cũ
    encoding_patterns = [
        r"Bieân",
        r"Tieán",
        r"Hoïc",
        r"Luaät",
        r"ñ",
    ]

    for pattern in encoding_patterns:
        if re.search(pattern, text, flags=re.IGNORECASE):
            issues.append("Nghi vấn lỗi mã hóa")
            suggestions.append(pattern)

    # 4. Kiểm tra dòng mục lục
    if re.search(r"\.{5,}", text):
        issues.append("Nghi vấn mục lục hoặc dấu chấm bất thường")

    return list(set(issues)), list(set(suggestions))


inspection = df["text_normalized"].apply(inspect_chunk)

df["loi_nghi_van"] = inspection.apply(
    lambda x: "; ".join(x[0])
)

df["vi_tri_nghi_van"] = inspection.apply(
    lambda x: "; ".join(x[1])
)

df_suspect = df[df["loi_nghi_van"] != ""].copy()

print("Số chunk nghi vấn:", len(df_suspect))
```

```python
def suggest_segmentation(text):
    try:
        # Underthesea phân tích cách tách từ
        result = word_tokenize(
            text,
            format="text"
        )
        return result
    except Exception:
        return ""


df_suspect["goi_y_tach_tu"] = df_suspect[
    "text_normalized"
].apply(suggest_segmentation)

# Giữ lại văn bản gốc để đối chiếu
df_review = df_suspect[
    [
        "document_id",
        "page",
        "chunk_id",
        "loi_nghi_van",
        "vi_tri_nghi_van",
        "text",
        "goi_y_tach_tu"
    ]
].copy()

df_review.head(10)
```

```python
import re
import pandas as pd
from google.colab import files

output_file = "kiem_tra_chunks.xlsx"

# Hàm loại bỏ các ký tự không hợp lệ trong Excel (các ký tự điều khiển ASCII ngoại trừ tab, newline, carriage return)
def remove_illegal_excel_chars(val):
    if not isinstance(val, str):
        return val
    # Loại bỏ các ký tự điều khiển không hợp lệ trong XML/Excel (0x00 - 0x08, 0x0B - 0x0C, 0x0E - 0x1F)
    val = re.sub(r'[\x00-\x08\x0B-\x0C\x0E-\x1F]', '', val)
    # Loại bỏ hoặc thay thế các ký tự Unicode PUA (Private Use Area) nếu cần
    # Ví dụ: một số ký tự đặc biệt gây lỗi khi export
    return val

# Áp dụng làm sạch dữ liệu cho cả 2 dataframe trước khi ghi
df_review_cleaned = df_review.map(remove_illegal_excel_chars) if hasattr(df_review, 'map') else df_review.applymap(remove_illegal_excel_chars)
df_cleaned = df.map(remove_illegal_excel_chars) if hasattr(df, 'map') else df.applymap(remove_illegal_excel_chars)

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    df_review_cleaned.to_excel(
        writer,
        sheet_name="Can_kiem_tra",
        index=False
    )

    df_cleaned.to_excel(
        writer,
        sheet_name="Toan_bo_chunks",
        index=False
    )

files.download(output_file)
```
